```python
import os
import json
```

```python
def get_lumi_dict(grl_csvs):
    lumi_dict = {}
    for grl in grl_csvs:
        with open(grl, 'r') as f:
            for i, line in enumerate(f):
                if i == 0: continue

                spline = line.split(",")
                lumi_rec = float(spline[3])
                run_num = int(spline[0])
                lumi_dict[run_num] = lumi_rec

    return lumi_dict
```

```python
with open("bad_runs.json", "r") as f:
    bad_runs_dict = json.load(f)
```

```python
bad_runs_dict_reform = {}
for entry in bad_runs_dict:
    for run in entry['runs']:
        bad_runs_dict_reform[run] = entry['reason']
```

```python
# lumi_dict = get_lumi_dict(["/cvmfs/faser.cern.ch/repo/sw/runlist/v6/faser_goodrunlist_2022_alps.csv", "/cvmfs/faser.cern.ch/repo/sw/runlist/v6/faser_goodrunlist_2023_alps.csv"])
lumi_dict = get_lumi_dict(["/cvmfs/faser.cern.ch/repo/sw/runlist/v8/faser_goodrunlist_2022_all.csv", "/cvmfs/faser.cern.ch/repo/sw/runlist/v8/faser_goodrunlist_2023_stable.csv"])
```

```python
unfiltered_lumi = sum(lumi_dict.values())
print(f"Lumi before filtering: {unfiltered_lumi/1000:.2f} /fb")
```

```python
bad_runs_list = []
for entry in bad_runs_dict:
    bad_runs_list += entry['runs']
```

```python
filtered_lumi = 0
for run, lumi in lumi_dict.items():
    if run in bad_runs_list:
        print(f"Run {run} in GRL and is a bad run! {bad_runs_dict_reform[run]}")
        continue
    if run < 8651:
        print(f"Run has no optical filter")
        continue
    filtered_lumi += lumi

print(f"Lumi after filtering: {filtered_lumi/1000:.2f} /fb")
```

```python
print(f"Difference in lumi: {(unfiltered_lumi - filtered_lumi)/1000:.2f} /fb")
```

```python
lumi_dict_v6 = get_lumi_dict(["/cvmfs/faser.cern.ch/repo/sw/runlist/v6/faser_goodrunlist_2022_alps.csv", "/cvmfs/faser.cern.ch/repo/sw/runlist/v6/faser_goodrunlist_2023_alps.csv"])
lumi_dict_v8 = get_lumi_dict(["/cvmfs/faser.cern.ch/repo/sw/runlist/v8/faser_goodrunlist_2022_all.csv", "/cvmfs/faser.cern.ch/repo/sw/runlist/v8/faser_goodrunlist_2023_stable.csv"])
```

```python
diff_good_runs = 0
lumi_tot_v6 = 0
lumi_tot_v8 = 0

for run, lumi_v8 in lumi_dict_v8.items():
    lumi_v6 = lumi_dict_v6.get(run, -1)
    
    # if lumi_v8 < 0: continue
    if run < 8651: continue
    if run in bad_runs_list: continue

    lumi_diff = lumi_v6 - lumi_v8

    # if lumi_diff != 0:
    #     print(f"Run {run}: Lumi v6 = {lumi_v6:.2f} /pb  Lumi v8 = {lumi_v8:.2f} /pb   Difference = {lumi_diff:.2f} /pb")
    lumi_tot_v6 += lumi_v6
    lumi_tot_v8 += lumi_v8    
    
    diff_good_runs += lumi_diff
```

```python
print(f"Lumi v6 = {lumi_tot_v6/1000:.2f} /pb")
print(f"Lumi v8 = {lumi_tot_v8/1000:.2f} /pb")
```

```python
unfiltered_lumi_v8 = sum(lumi_dict_v8.values())
print(f"Difference in v6 and v8 lumis {(unfiltered_lumi - unfiltered_lumi_v8)/1000:.2f} /fb")
```

```python
print(f"Diff between good runs = {diff_good_runs/1000:.2f} fb")
```
